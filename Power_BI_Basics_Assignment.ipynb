# Question 1: What is Power BI and why is it used in businesses?

## Answer:

Power BI is a business intelligence and data visualization tool developed by Microsoft. It helps users connect to different data sources, analyze data, and create interactive dashboards and reports.

Businesses use Power BI to convert raw data into meaningful insights for better decision-making.

### Why Power BI is used in businesses:

1. Data Visualization  
   Power BI creates charts, graphs, and dashboards that make data easy to understand.

2. Better Decision-Making  
   Businesses can analyze performance and make informed decisions.

3. Real-Time Reporting  
   It provides live dashboards and real-time updates.

4. Data Integration  
   Power BI can connect to Excel, SQL databases, cloud services, and many other data sources.

5. Improved Productivity  
   Automated reports save time and reduce manual work.

### Conclusion:

Power BI helps businesses analyze data efficiently and make smarter business decisions.

# Question 2: Name and explain the three main components of Power BI.

## Answer:

The three main components of Power BI are:

### 1. Power BI Desktop

Power BI Desktop is a Windows application used to create reports and dashboards.

#### Functions:
- Import data
- Clean and transform data
- Create visualizations and reports

### 2. Power BI Service

Power BI Service is a cloud-based online platform.

#### Functions:
- Publish reports online
- Share dashboards with teams
- Access reports from anywhere


### 3. Power BI Mobile

Power BI Mobile is a mobile application used on smartphones and tablets.

#### Functions:
- View reports on mobile devices
- Monitor dashboards anytime
- Get real-time notifications

### Conclusion:

These components work together to help users create, share, and access reports easily.


# Question 3: Explain the Power BI workflow.

## Answer:

The Power BI workflow is the step-by-step process of collecting, cleaning, analyzing, and visualizing data.

## Steps in Power BI Workflow:

### 1. Data Collection
Data is collected from sources like Excel, SQL databases, websites, and cloud services.

### 2. Data Cleaning and Transformation
Using Power Query, data is cleaned by:
- Removing duplicates
- Handling missing values
- Changing data types

### 3. Data Modeling
Relationships are created between tables to organize the data properly.

### 4. Data Visualization
Charts, graphs, KPIs, and dashboards are created.

### 5. Report Publishing
Reports are published to Power BI Service.

### 6. Sharing and Monitoring
Users can access reports through web browsers or mobile apps.

### Conclusion:

The Power BI workflow helps convert raw data into useful business insights.

# Question 6: Define Data View, Report View, and Model View. Explain the purpose of each view.

## Answer:

Power BI provides different views to manage and analyze data efficiently.

## 1. Data View

### Definition:
Data View displays the complete dataset in a table format.

### Purpose:
- View rows and columns
- Check data quality
- Analyze raw data
- Create calculated columns


## 2. Report View

### Definition:
Report View is used to create charts, graphs, dashboards, and reports.

### Purpose:
- Create visualizations
- Design dashboards
- Add filters and slicers
- Perform business analysis


## 3. Model View

### Definition:
Model View shows relationships between different tables in the dataset.

### Purpose:
- Create relationships
- Manage data models
- Define primary and foreign keys
- Improve data organization

### Conclusion:

Each view in Power BI serves a different purpose:
- Data View → Data analysis
- Report View → Visualization
- Model View → Relationship management

# Question 8: Split Owner Name to create two new columns as First Name and Last Name

## Answer:

### Steps to Split Owner Name in Power BI:

### Step 1:
Open Power Query Editor.

### Step 2:
Select the “Owner Name” column.

### Step 3:
Go to the:
### Transform Tab → Split Column

### Step 4:
Choose:
### “By Delimiter”

### Step 5:
Select:
### Space (“ ”) as delimiter

### Step 6:
Choose:
### Split at Left-most Delimiter

### Step 7:
Rename the new columns as:
- First Name
- Last Name


### Example:

| Owner Name | First Name | Last Name |
|------------|------------|------------|
| Rahul Sharma | Rahul | Sharma |
| Neha Verma | Neha | Verma |

Splitting columns helps improve data organization and makes analysis easier in Power BI.
